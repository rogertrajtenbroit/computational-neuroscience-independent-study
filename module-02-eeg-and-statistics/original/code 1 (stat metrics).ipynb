```python
import numpy as np

# define list
values = [3, 15, 7, 8, 11, 12, 6, 3, 2, 0, 13, 11, 9, 12, 11, 6, 8, 5]

# compute stat metrics
mean_value = np.mean(values)  # you can't use the word mean because "mean" is already a function. same with median and range
median_value = np.median(values)
data_range = np.max(values) - np.min(values)
standard_deviation = np.std(values)

# 3. Print the results to the console
print(f"Mean: {mean_value:.2f}") 
print(f"Median: {median_value}")
print(f"Range: {data_range}")
print(f"Standard Deviation: {standard_deviation:.2f}")

#The f that goes in print(f" lets Python know that inside the curly braces { there are variables or calculations that it has to process before printing.
```
