```python
import numpy as np
import matplotlib.pyplot as plt

# haces la random sequence de numeros.
ranvar = np.random.uniform(0, 100, 10000)    # how to use this: np.random.uniform(low, high, size)
print(ranvar)
```

```python
# ahora haces el histograma: plt.hist(datos, bins, color, edgecolor)
 
plt.hist(ranvar, bins=30, color='skyblue', edgecolor='black')

# la data que aparece arriba del grafico la da la funcion de plt.hist.  1) la altura de cada barra 2)los bordes de cada barra 3) cuantas barras hizo
# si la queres sacar podes poner un ; al final de la funcion, justo despues de cerrar el parentesis
```

```python

import scipy.stats as stats

# 1. Configure the 2x2 layout matrix for subplots
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# --- Panel 1: Homogeneous (Uniform) Distribution ---
x_uni = np.linspace(-10, 110, 500)
y_uni = stats.uniform.pdf(x_uni, loc=0, scale=100)
axs[0, 0].plot(x_uni, y_uni, color='blue', linewidth=2)
axs[0, 0].set_title('Homogeneous Distribution (Theoretical)')

# --- Panel 2: Normal (Gaussian) Distribution ---
x_norm = np.linspace(0, 100, 500)
y_norm = stats.norm.pdf(x_norm, loc=50, scale=15)
axs[0, 1].plot(x_norm, y_norm, color='green', linewidth=2)
axs[0, 1].set_title('Normal Distribution (Theoretical)')

# --- Panel 3: Log-Normal Distribution ---
x_log = np.linspace(0.1, 50, 500)
y_log = stats.lognorm.pdf(x_log, s=0.5, scale=np.exp(2))
axs[1, 0].plot(x_log, y_log, color='red', linewidth=2)
axs[1, 0].set_title('Log-Normal Distribution (Theoretical)')

# --- Panel 4: Poisson Distribution ---
x_pois = np.arange(0, 15)
y_pois = stats.poisson.pmf(x_pois, mu=4)
axs[1, 1].step(x_pois, y_pois, where='mid', color='purple', linewidth=2, marker='o')
axs[1, 1].set_title('Poisson Distribution (Theoretical)')

# Automatically adjust padding and spacing between subplots
plt.tight_layout()
plt.show()
```

```python
# It resembles a homogeneous distribution more closely because a homogeneous distribution is what you get when all outcomes have the same
#probability of occurring. That is exactly what happens when you generate a RANDOM sequence of numbers from 1 to 100 and the length
#of the sequence (i.e., the number of observations, n) is sufficiently large.
```
