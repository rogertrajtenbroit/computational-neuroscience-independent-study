```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([20.0, 16.0, 19.8, 18.4, 17.1, 15.5, 14.7, 17.1, 15.4, 16.2, 15.0, 17.2, 16.0, 17.0, 14.4])
y = np.array([88.6, 71.6, 93.3, 84.3, 80.6, 75.2, 69.7, 82.0, 69.4, 83.3, 79.6, 82.6, 80.6, 83.5, 76.3])

xmean = np.mean(x)
ymean = np.mean(y)

a, b = np.polyfit(x, y, 1)

print("xmean =", xmean)
print("ymean =", ymean)
print("a =", a)
print("b =", b)

plt.scatter(x, y, color='skyblue', edgecolor='black', label='datos')
plt.plot(x, a*x + b, color='red', label='least-square line',)
plt.scatter(xmean, ymean, color='black', zorder=3, label='(xmean, ymean)')
plt.xlabel('chirps per second')
plt.ylabel('temperature (F)')
plt.legend()
plt.show()
```
