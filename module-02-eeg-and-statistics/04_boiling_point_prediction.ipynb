# Boiling point prediction

*Module 2 · Computational Neuroscience Independent Study*

I fit a straight line through two altitude and boiling-point values and use it to estimate the boiling point at Everest’s altitude. This is an exercise in extrapolation from two points.

[Module overview](README.md) · [Execution notes](VERIFICATION.md)

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the experimental data points as NumPy arrays
# x: Altitude in kilometers (Control/Independent variable)
# y: Boiling point in °C (Response/Dependent variable)
x_points = np.array([0.0, 2.0])
y_points = np.array([100.0, 93.0])

# 2. Compute the linear function coefficients (y = ax + b)
a, b = np.polyfit(x_points, y_points, 1)
# Degree=1 means a straight line. 
# It automatically calculates the slope (a) and intercept (b) from the two points.

# 3. Predict the boiling point on Mount Everest (x = 8.85 km)
everest_altitude = 8.85
everest_boiling_point = a * everest_altitude + b

print(f"1) Control variable (x): Altitude (km)")
print(f"   Response variable (y): Boiling Point (°C)")
print(f"2) Linear function equation: y = {a:.2f}x + {b:.2f}")
print(f"3) Predicted boiling point on Mount Everest ({everest_altitude} km): {everest_boiling_point:.2f}°C")

plt.scatter(x_points, y_points, color='skyblue', edgecolor='black', label='datos')
plt.plot(x_points, a * x_points + b , color='red', label='least-square line',)
plt.legend()
plt.show()
```
